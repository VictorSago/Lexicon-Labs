## Filter rows

How many listings are entire homes?

mask

```python
listings["room_type"] == "Entire home/apt"
```

```python
mask = listings["room_type"] == "Entire home/apt"
print(mask.sum())
print(mask.mean())
```

mask as a filter

```python
listings[mask].shape
```

Which listings are in Södermalm, and also have room for four or more guests?

```python
listings[listings["neighbourhood_cleansed"] == "Södermalms" and listings["accommodates"] >= 4]
```

```python
listings[(listings["neighbourhood_cleansed"] == "Södermalms") & (listings["accommodates"] >= 4)]
```

```python
central = ["Södermalms", "Östermalms", "Norrmalms"]

print(listings["neighbourhood_cleansed"].isin(central).sum())
print(listings["accommodates"].between(2, 4).sum())
print((~mask).sum())
print(listings["review_scores_rating"].isna().sum())
```

## Sort and pick rows and columns

```python
view = ["name", "neighbourhood_cleansed", "room_type", "accommodates", "number_of_reviews", "review_scores_rating"]
listings[view].sort_values("number_of_reviews", ascending=False).head(5)
```

```python
listings[view].sort_values(["accommodates", "number_of_reviews"], ascending=[False, False]).head(5)
```

```python
listings.nlargest(5, "accommodates")[view]
```

```python
top = listings[view].sort_values("number_of_reviews", ascending=False).head(5)
print(top.loc[825, "name"])
print(top.iloc[0]["name"])
```
