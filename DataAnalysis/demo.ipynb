### What is in the data?

The 84 columns are roughly of these kinds:

- **About the listing:** the name, the type of home (`property_type`, `room_type`), how many guests it fits (`accommodates`), and the number of bedrooms and beds.
- **Where it is:** the neighbourhood (`neighbourhood_cleansed`, 14 areas in Stockholm) and the coordinates.
- **The price and the rules:** the price per night (`price`) and the minimum number of nights (`minimum_nights`).
- **Availability:** how many days the listing can be booked, for example during the next year (`availability_365`).
- **Reviews:** how many reviews the listing has (`number_of_reviews`), the rating (`review_scores_rating`), and the dates of the first and last review.
- **The host:** for example whether the host is a superhost (`host_is_superhost`).
- **Technical columns:** links, ID numbers and the date of the collection. We will not use them much.

Some of the columns are **completely empty**. We will come back to those on Monday next week, when we clean this table.

Over the coming days we will mostly use a handful of columns: `room_type`, `neighbourhood_cleansed`, `accommodates`, `bedrooms`, `beds`, `price`, `number_of_reviews`, `review_scores_rating`, `availability_365` and `minimum_nights`. In the labs you will get the chance to explore the **other** columns on your own, so if something catches your eye today, write it down.

Which listings are in Södermalm, and also have room for four or more guests?

```python
listings[listings['neighbourhood_cleansed'] == 'Södermalms' and listings['accommodates'] >= 4]
```

```python
listings[(listings['neighbourhood_cleansed'] == 'Södermalms') & (listings['accommodates'] >= 4)]

# & means AND, | means OR, ~ means NOT.
```

- `isin()`: is the value **in a list**? It is much shorter than writing one condition for every value and joining them with `|`.
- `between()`: is a number **between two values**?
- `~`: the **not** sign. It turns every `True` into `False` and the other way around.
- `isna()`: is the value **missing**?

```python
central = ['Södermalms', 'Östermalms', 'Norrmalms']

print(listings['neighbourhood_cleansed'].isin(central).sum())      
print(listings['accommodates'].between(2, 4).sum())               
print((~mask).sum())                                              
print(listings['review_scores_rating'].isna().sum())
```

**In Södermalm, among entire homes for four or more guests: which five listings have the most reviews, and how are they rated?**

```python
answer = (
    listings[
        (listings['neighbourhood_cleansed'] == 'Södermalms')
        & (listings['room_type'] == 'Entire home/apt')
        & (listings['accommodates'] >= 4)
    ]
    .sort_values('number_of_reviews', ascending=False)
    .head(5)[['name', 'accommodates', 'number_of_reviews', 'review_scores_rating']]
)

answer
```
